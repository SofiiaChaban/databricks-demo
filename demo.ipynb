```python
import dbldatagen as dg

spec = (dg.DataGenerator(spark, rows=50_000_000, partitions=8)
    .withColumn("order_id", "long", uniqueValues=50_000_000)
    .withColumn("customer_id", "string", template=r"CUST_\d{6}")
    .withColumn("amount", "decimal(10,2)", minValue=1, maxValue=1000, random=True)
    .withColumn("order_date", "date", 
                begin="2025-04-20", end="2026-04-17", random=True)
    .withColumn("status", "string", 
                values=["completed", "pending", "cancelled"], 
                weights=[70, 25, 5], random=True)
)
# display(spec)

df = spec.build()
display(df)

df.write.mode("overwrite").saveAsTable("demo.test_db.bronze_orders")
```
